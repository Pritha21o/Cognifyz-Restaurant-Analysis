## Task 2: Restaurant Recommendation System

### Objective
To recommend restaurants based on user preferences such as cuisine, city, and price range.

### Approach
This project uses a rule-based, content-based filtering approach. It filters restaurants according to the user's selected cuisine, city, and price range, then sorts matching restaurants by aggregate rating.

The system does not train a machine-learning model. It uses predefined filtering rules and restaurant attributes to generate recommendations.

```python
import pandas as pd
from pathlib import Path

folder = Path(
    r"C:\Users\prith\Restaurant-Data-Analysis\data"
)

# Find the CSV automatically
csv_files = list(folder.glob("*.csv"))

if csv_files:
    df = pd.read_csv(csv_files[0])

    df_clean = df.copy()
    df_clean["Cuisines"] = df_clean["Cuisines"].fillna("Unknown")

    display(df_clean.head())
    print("Dataset loaded:", df_clean.shape)
else:
    print("CSV file not found. Check your data folder.")
```

```python

# Check the columns available for recommendations
print(df_clean.columns.tolist())

# Preview a few restaurants
df_clean[
    [
        "Restaurant Name",
        "City",
        "Cuisines",
        "Price range",
        "Aggregate rating"
    ]
].head()
```

```python

# Define sample user preferences
preferred_cuisine = "Japanese"
preferred_price = 3
preferred_city = "Makati City"

print("Cuisine preference:", preferred_cuisine)
print("Price range:", preferred_price)
print("Preferred city:", preferred_city)
```

```python

# Check which restaurants match all three preferences

matching_restaurants = df_clean[
    df_clean["Cuisines"].str.split(",").apply(
        lambda cuisines: preferred_cuisine.lower()
        in [c.strip().lower() for c in cuisines]
    )
    & (df_clean["Price range"] == preferred_price)
    & (df_clean["City"] == preferred_city)
]

matching_restaurants[
    ["Restaurant Name", "City", "Cuisines", "Price range", "Aggregate rating"]
]
```

```python
ranked_restaurants = matching_restaurants.sort_values(
    by="Aggregate rating",
    ascending=False
)

ranked_restaurants[
    ["Restaurant Name", "City", "Cuisines",
     "Price range", "Aggregate rating", "Votes"]
]
```

```python
# Test a different set of user preferences

test_cuisine = "Japanese"
test_price = 4
test_city = "Mandaluyong City"

test_recommendations = df_clean[
    df_clean["Cuisines"].str.split(",").apply(
        lambda cuisines: test_cuisine.lower()
        in [c.strip().lower() for c in cuisines]
    )
    & (df_clean["Price range"] == test_price)
    & (df_clean["City"] == test_city)
]

test_recommendations = test_recommendations.sort_values(
    by="Aggregate rating",
    ascending=False
)

test_recommendations[
    ["Restaurant Name", "City", "Cuisines",
     "Price range", "Aggregate rating", "Votes"]
].head(10)
```

```python
test_cuisine = "Mexican"
test_price = 1
test_city = "Makati City"

no_match_recommendations = df_clean[
    df_clean["Cuisines"].str.split(",").apply(
        lambda cuisines: test_cuisine.lower()
        in [c.strip().lower() for c in cuisines]
    )
    & (df_clean["Price range"] == test_price)
    & (df_clean["City"] == test_city)
]

print("Number of recommendations:", len(no_match_recommendations))
```

```python
if no_match_recommendations.empty:
    print("No matching restaurants found.")
    print("Try changing your cuisine, price range, or city.")
else:
    print("Matching restaurants found:")
    display(no_match_recommendations)
    
```

## Limitations

* Recommendations depend on the restaurant data available in the dataset.
* Results are limited by the user's selected cuisine, city, and price range.
* The system does not learn from user behavior or update its recommendations through model training.
* If no restaurants match the selected preferences, the system may return no recommendations.
